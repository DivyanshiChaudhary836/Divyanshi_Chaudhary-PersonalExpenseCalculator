## 2. Load your dataset

```python
df = pd.read_csv("Student_18_dataset.csv")
display(df.head())
```

## 3. Explore the data
Use `shape`, `columns`, `info()`, `isnull().sum()` and `duplicated().sum()` to inspect the dataset.


```python
# TODO: Display the shape and column names
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())

# TODO: Inspect data types and missing values
print("\nDataFrame Info: ")
df.info()
print("\nMissing values: ")
print(df.isnull().sum())

# TODO: Count duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())
```

## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
# TODO: Select numeric columns relevant to your project
# TODO: Convert them to a NumPy array using .to_numpy()
amount_arr=df["Amount"].to_numpy()

# TODO: Calculate average, maximum and minimum
avg_amount=np.mean(amount_arr)
max_amount=np.max(amount_arr)
min_amount=np.min(amount_arr)
print(f"Average expense: {avg_amount:.2f}")
print(f"Maximum expense: {max_amount:.2f}")
print(f"Minimum expense: {min_amount:.2f}")
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# TODO: Create calculated column(s)
#1- to find out what percent of the total share is the given amount
total_expense=df["Amount"].sum()
df["share_of_total"]=(df["Amount"]/total_expense)*100
#2- to find out whether the amount spend is more, less or equal to the average amount
df["Expense_level"]=np.where(df["Amount"]>= avg_amount, "Above Average", "Below Average")

# TODO: Filter records using a condition
high_expenses=df[df["Amount"]>avg_amount]
print(f"Number of above-average transactions: {len(high_expenses)}")
display(high_expenses.head())

# TODO: Sort records and display the top/bottom results
sorted=df.sort_values(by="Amount",ascending=False)
print("Top 5 highest expenses are: ")
display(sorted.head())
print("Bottom 5 lowest expenses are: ")
display(sorted.tail())
```

## 6. Your individual project challenge
**Challenge:** Calculate total expenses; category-wise totals and averages; identify largest expense category.

Write your solution below and explain what the result means.


```python
# TODO: Solve your assigned challenge
#1- total expenses 
total_expenses = df["Amount"].sum()
print(f"Total Expenses: {total_expenses:.2f}")
#2- category-wise totals
category_wise=df.groupby("Category")["Amount"].sum()
print(category_wise)
#3- category wise averages
category_wise=df.groupby("Category")["Amount"].mean()
print(category_wise)
#4- largest expense category
largest_category=category_wise.idxmax()
largest_amount=category_wise.max()
print(f"The largest expense category is- {largest_category} and the maximum amount spent on it is- {largest_amount}")
```
